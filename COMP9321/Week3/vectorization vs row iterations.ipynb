```python
import pandas as pd
import numpy as np

#example context: you have transaction data and you need :

# Add 10% tax

# Flag high-value transactions (> 1000)

# Apply tiered discount logic


df = pd.DataFrame({
    'transaction ID' : [23420, 34250, 113430, 89290, 12788],
    'amount': [200, 1500, 300, 2500, 700]
    
})

df
```

```python
# iterrows avoid this approach as it will cause perforamnce hit for large datasets

for index, row in df.iterrows():
    df.loc[index, 'taxed'] = row['amount'] * 1.10
    
df
```

```python


df['taxed'] = df['amount'].apply(lambda x: x * 1.10)

df
```

```python
# apply can be helpful if the logic is not easy to vectorize
# or if you have multiple comlumns and you need to apply some business logic

def discount(row):
    if row['amount'] > 2000:
        return row['amount'] * 0.90
    elif row['amount'] > 1000:
        return row['amount'] * 0.95
    else:
        return row['amount']

df['discounted'] = df.apply(discount, axis=1)

df
```

```python
# vectorized is faster, more effcient (as it avoids interpreter overhead and leverge optimized memory access)

df['taxed'] = df['amount'] * 1.10
df['high_value'] = (df['amount'] > 1000).astype(int)

df
```
